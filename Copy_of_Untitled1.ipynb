```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. تحويل السعر لرقم (ده اللي هيظبط المحور X)
df['price'] = pd.to_numeric(df['price'], errors='coerce')

# 2. تحديد حجم الرسمة عشان متبقاش مضغوطة
plt.figure(figsize=(12, 6))

# 3. رسم العلاقة مع جعل النقاط شفافة قليلاً (alpha) عشان زحمة البيانات تبان
sns.scatterplot(data=df, x="price", y="num_subscribers", alpha=0.6, color='skyblue')

# 4. إضافة عناوين واضحة
plt.title("Effect of Price on Number of Subscribers", fontsize=15)
plt.xlabel("Price", fontsize=12)
plt.ylabel("Number of Subscribers", fontsize=12)

# 5. إضافة شبكة بسيطة لتسهيل القراءة
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

```python
df.groupby("level")["num_reviews"].sum().sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt

# تجميع البيانات
level_reviews = df.groupby("level")["num_reviews"].sum().sort_values(ascending=False)

# الرسم
plt.figure(figsize=(8,5))
level_reviews.plot(kind='bar')

plt.title("Number of Reviews by Level")
plt.xlabel("Level")
plt.ylabel("Total Reviews")

plt.xticks(rotation=45)
plt.show()
```

```python
df.groupby("subject")["num_reviews"].sum().sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt

# تجميع البيانات
subject_reviews = df.groupby("subject")["num_reviews"].sum().sort_values(ascending=False)

# الرسم
plt.figure(figsize=(10,5))
subject_reviews.plot(kind='barh')

plt.title("Number of Reviews by Subject")
plt.xlabel("Total Reviews")
plt.ylabel("Subject")

plt.show()
```

```python

df["published_timestamp"] = pd.to_datetime(df["published_timestamp"], utc=True, errors='coerce')

# ك
df["year"] = df["published_timestamp"].dt.year
result = df.groupby("year")["num_reviews"].sum()

print(result)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# تحويل لـ DataFrame
year_reviews = result.reset_index()

# الرسم
plt.figure(figsize=(12,6))

sns.lineplot(
    data=year_reviews,
    x="year",
    y="num_reviews",
    marker="o"
)

plt.title("Reviews Growth Over Years")
plt.xlabel("Year")
plt.ylabel("Total Reviews")

plt.grid(True)

plt.show()
```
